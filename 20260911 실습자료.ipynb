```python
#두 자연수 10과 5가 주어진다.
#이때, A+B, A-B, A*B, A/B(몫), A%B(나머지)를 출력하는 프로그램을 작성하시오.

A = 10
B = 5

print(A+B)
print(A-B)
print(A*B)
print(A/B)
print(A%B)
```
